```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 6])
y = np.array([10, 250])

plt.plot(x, y)
plt.show()
```

If we do not specify the points on the x-axis, they will get the default values 0, 1, 2, 3 etc., depending on the length of the y-points.

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, marker = 'X') # Marker give a type of mark to each point
plt.show()
```

**Types of markers**  <br>
'o'	Circle <br>
'*'	Star <br>
'.'	Point <br>
','	Pixel <br>
'x'	X <br>
'X'	X (filled) <br>
'+'	Plus <br>
'P'	Plus (filled) <br>
's'	Square <br>
'D'	Diamond <br>
'd'	Diamond (thin) <br>

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, linestyle="dotted", markerfacecolor="red",markeredgecolor="green", marker="d", color="blue", )
plt.show()
```

**Line types** <br>
'-'	Solid line <br>
':'	Dotted line <br>
'--'	Dashed line <br>
'-.'	Dashed/dotted line <br>

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([1,8, 27, 64])

plt.plot(ypoints, marker = 'X', c='y') # Marker give a type of mark to each point. c='colorcode' specifies the color of the line
plt.show()
```

## **Shortcut string notation parameter.** <br>
Called as fmt <br>
*marker|line|color* <br>
Ex. <br>
1. plt.plot(ypoints, '*--y')  * specifies the star marker, -- specifies the line type and y specifies the yellow color.

```python
import matplotlib.pyplot as plt
import numpy as np

x= np.array([2023, 2024, 2025, 2026])
y1 = np.array([15,25,30,20])
y2 = np.array([17,23,38, 5])
y3= np.array([13,15,20, 30])

plt.title("Class size over the years", fontsize=20, family="serif", fontweight="bold")
plt.xlabel("Year", fontsize=15)
plt.ylabel("Class size", fontsize=15)

plt.xticks(x) # Sets the x valus on the x-axis to x and controls the increment

plt.plot(x, y1)
plt.plot(x,y2)
plt.plot(x,y3)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Bar chart = compare categories of data by representing each category with

categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy",])
values = np.array([4, 3, 2, 5, 3, ])

plt.figure(figsize=(7,5)) # Reconfigures the graph to make it 7 units x 5 units
plt.bar(categories,values, color="#a87132") # to make a horizontal bar chart, use plt.barh()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x1 = [0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8] # Hours studied
y1 = [55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87] # Grades

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) # Hours studied
y2 = np.array([50,58, 65, 70,72, 78,83, 88, 92, 95, 97]) # Grades

plt.scatter(x1, y1, s=50, color="#d11d59", label="Class A") # s defines the size of the marker
plt.scatter(x2, y2, s=50, color="#6ba867", label="Class B")

plt.xlabel("Hours studied")
plt.ylabel("Test Scores")

plt.axhline(y=85, color="#4287f5", linestyle="--", label="Marks to get A grade")

plt.legend()
plt.show()
```
